## Popularity Based Movie Recommendation System

```python
import kagglehub
path = kagglehub.dataset_download("tmdb/tmdb-movie-metadata")
```

```python
import os
os.listdir(path)
```

```python
import pandas as pd
credits=pd.read_csv(os.path.join(path, "tmdb_5000_credits.csv"))
movies=pd.read_csv(os.path.join(path, "tmdb_5000_movies.csv"))
```

```python
movies.sample()
```

```python
movies=movies[['title','popularity','vote_average','vote_count']]
movies.head()
```

```python
movies.info()
```

```python
movies.describe()
```

```python
movies['title'].is_unique
```

```python
movies['title'].value_counts()
```

```python
m=movies['vote_count'].quantile(0.75)
movies['weighted_votes'] = (
    (movies['vote_count'] * movies['vote_average'])
    / (movies['vote_count'] + m)
    +
    (m * movies['vote_average'].median())
    / (m + movies['vote_count'])
)
```

```python
movies.sample()
```

```python
movies.drop(columns=['vote_average','vote_count'],inplace=True)
```

```python
movies=movies.groupby('title')[['popularity','weighted_votes']].mean().reset_index()
```

```python
movies=movies.sort_values(['weighted_votes','popularity'],ascending=False)
```

```python
movies.sample()
```

```python
movies.to_csv('movies.csv',index=False)
```
